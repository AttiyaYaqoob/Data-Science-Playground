# STEP 8 DATA SPLIT

```python
df.shape
```

```python
threshold = 0.5 # 50% missing allowed

cols_to_drop = []

for col in df.columns:
    missing_ratio = df[col].isnull().mean()
    if missing_ratio > threshold:
        cols_to_drop.append(col)

df.drop(columns=cols_to_drop, inplace=True)

print("Dropped due to missing values:", cols_to_drop)
```

```python

row_threshold = 0.5  # 50% missing in a row

df = df[df.isnull().mean(axis=1) < row_threshold]

print("Rows removed with high missing values")
```

# Find Loss and Accuracy
# Results on Training Metrics

```python
from sklearn.metrics import ( mean_absolute_error, mean_squared_error, r2_score,mean_absolute_percentage_error)
```

```python
# how we predict the accuracy of model trained?
r2 = r2_score( Y_TRAIN, training_data_prediction) #Y_TRAIN are original answers while "training_data_prediction" are model answers
mae = mean_absolute_error(Y_TRAIN,training_data_prediction)
mse = mean_squared_error(Y_TRAIN,training_data_prediction)
rmse =  np.sqrt(mse)
print(f"R^2  : {r2:.4f}")   # called accuracy
print(f"MAE  : {mae:.4f}")   #loss
print(f"MSE  : {mse:.4f}")   #loss
print(f"RMSE  : {rmse:.4f}")   #loss
```
