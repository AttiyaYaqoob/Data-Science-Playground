Step 2: Download NLTK Data

```python
#Natural Language Toolkit (NLTK) that splits a string of text into individual words, numbers, and punctuation marks (tokens).
from nltk.tokenize import word_tokenize
# commonly used words (e.g., "the," "is," "and," "in") that are filtered out in NLP to reduce data noise and improve processing efficiency
from nltk.corpus import stopwords
# Tool in the NLTK (Natural Language Toolkit) library used to find the base or dictionary form of a word, known as a lemma
from nltk.stem import WordNetLemmatizer
```

Load Dataset

```python
text = '''Natural Language Processing (NLP) is a field of Artificial Intelligence. It helps computers understand human language!'''
```

Step 4: Text Cleaning Function

```python
text = text.lower()
print(text)
```

```python
tokens = word_tokenize(text)
print(tokens)
```

```python
tokens = [word for word in tokens if word.isalnum()]
print(tokens)
```

Remove stop words

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words]
print(filtered_tokens)
```

Step 6: Stemming (Reduce words to root form)

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
print(stemmed_words)
```

Step 7: Lemmatization (Better than stemming)

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word)
for word in filtered_tokens]
print(lemmatized_words)
```
