```python
import pandas as pd
import numpy as np

df = pd.read_csv('house_data.csv')

# Step 1: Drop columns with more than 50% missing values (as per previous logic)
df.dropna(thresh=len(df) * 0.5, axis=1, inplace=True)

# Step 2: Fill numerical NaN values with the mean
for col in df.columns:
    if df[col].dtype == 'float64' and df[col].isna().sum() != 0:
        df[col] = df[col].fillna(df[col].mean())

# Step 3: Drop 'FireplaceQu' column (as per previous logic)
if 'FireplaceQu' in df.columns:
    df = df.drop('FireplaceQu', axis=1)

# Step 4: Fill categorical NaN values with the mode
for col in df.columns:
    if df[col].dtype == 'object' and df[col].isna().sum() != 0:
        df[col] = df[col].fillna(df[col].mode()[0])

# Step 5: Clean column names by stripping whitespace
df.columns = df.columns.str.strip()

print("DataFrame after cleaning:")
df.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select only numerical columns for correlation calculation
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Calculate the correlation matrix
corr_matrix = df[numerical_cols].corr()

# Set up the matplotlib figure
plt.figure(figsize=(18, 15))

# Draw the heatmap
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Numerical Features', fontsize=20)
plt.show()
```
