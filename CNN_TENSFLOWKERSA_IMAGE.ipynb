```python
(X_TRAIN,Y_TRAIN),(X_TEXT,Y_TEST)=tf.keras.datasets.mnist.load_data()
```

DATA PREPROCESSING NORMALIZE (0-255->0-1)

```python
X_TRAIN=X_TRAIN/255.0
X_TEXT=X_TEXT/255.0
```

DNN **bold text** FLATER LAYER,FLODER NAME CLASS K ANME HOTA HA

```python
X_TRAIN_FLATEN=X_TRAIN.reshape(len(X_TRAIN),28*28)
X_TEXT_FLATEN=X_TEXT.reshape(len(X_TEXT),28*28)
```

```python
type(X_TEXT_FLATEN)
```

```python
X_TRAIN_FLATEN.shape
```

```python
X_TEXT_FLATEN.shape
```

BUILD DNNN MODEL

```python
dnn_model=models.Sequential([
    layers.Dense(128,activation='relu',input_shape=(X_TEXT_FLATEN.shape[1],)),
    layers.Dense(64,activation='relu'),

    layers.Dense(10,activation='softmax')
])
```

```python
dnn_model.summary()
```

npne model

```python
dnn_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history_dnn=dnn_model.fit(X_TRAIN_FLATEN,Y_TRAIN,epochs=10,batch_size=32,validation_split=0.2)
```

TESTING

```python
test_loss,test_accuracy=dnn_model.evaluate(X_TEXT_FLATEN,Y_TEST)
print("DNN TEXT ACCURANCE",test_accuracy)
```

```python
pred=dnn_model.predict(X_TEXT_FLATEN)
print(pred[0])
```

```python
import numpy as np

data_predication=np.argmax(pred,axis=1)
data_predication
```
