```python

# Drop columns with >50% missing (not applicable here as missing values are only 2%)
# For numerical columns, we'll impute with the mean.
for col in df.select_dtypes(include=np.number).columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mean())

# For categorical columns, we'll impute with the mode.
for col in df.select_dtypes(include='object').columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after imputation:")
print(df.isnull().sum())
```

## Step 4: Perform OLS Regression

```python
import statsmodels.formula.api as smf

# Define the OLS model formula
# All numerical columns except 'stress_level' itself are used as independent variables
formula = 'stress_level ~ age + experience_years + daily_work_hours + sleep_hours + caffeine_intake + bugs_per_day + commits_per_day + meetings_per_day + screen_time + exercise_hours'

# Create and fit the OLS model
ols_model = smf.ols(formula=formula, data=df).fit()

# Print the summary of the regression results
print(ols_model.summary())
```

## Step 5: Calculate Feature Importance with Random Forest Regressor

```python
from sklearn.ensemble import RandomForestRegressor

# Prepare the data for Random Forest
# Features (X) are all numerical columns except 'stress_level'
feature_cols = ['age', 'experience_years', 'daily_work_hours', 'sleep_hours',
                'caffeine_intake', 'bugs_per_day', 'commits_per_day',
                'meetings_per_day', 'screen_time', 'exercise_hours']
X = df[feature_cols]
y = df['stress_level']

# Initialize and train the Random Forest Regressor model
rf_model = RandomForestRegressor(n_estimators=100, random_state=42) # n_estimators is the number of trees in the forest
rf_model.fit(X, y)

# Get feature importances
feature_importances = pd.DataFrame({
    'feature': feature_cols,
    'importance': rf_model.feature_importances_
}).sort_values(by='importance', ascending=False)

print("Feature Importances (Random Forest Regressor):")
display(feature_importances)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plotting feature importances
plt.figure(figsize=(12, 7))
sns.barplot(x='importance', y='feature', data=feature_importances)
plt.title('Feature Importance from Random Forest Regressor')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

## Step 6: Predict Stress Level for New Input Features

```python
import pandas as pd

# Define new input features for prediction
# Ensure the order and names of columns match the training data
new_features = pd.DataFrame({
    'age': [30.0],
    'experience_years': [5.0],
    'daily_work_hours': [9.0],
    'sleep_hours': [6.0],
    'caffeine_intake': [3.0],
    'bugs_per_day': [10.0],
    'commits_per_day': [15.0],
    'meetings_per_day': [4.0],
    'screen_time': [12.0],
    'exercise_hours': [1.0]
})

# Make a prediction using the trained Random Forest Regressor model
predicted_stress_level = rf_model.predict(new_features)

print(f"Predicted Stress Level for the new features: {predicted_stress_level[0]:.2f}")
```

## Step 7: Sensitivity Analysis on Daily Work Hours

```python
# Create a range of daily_work_hours for sensitivity analysis
daily_work_hours_range = np.arange(4.0, 13.0, 1.0) # From 4 to 12 hours, increment by 1

# Prepare an empty list to store predictions
sensitivity_predictions = []

# Base features from the previous prediction (excluding daily_work_hours)
base_features = new_features.drop(columns=['daily_work_hours']).iloc[0]

for hours in daily_work_hours_range:
    # Create a new DataFrame for each hour value
    temp_features = pd.DataFrame([base_features.to_dict()])
    temp_features['daily_work_hours'] = hours

    # Reorder columns to match the training data
    temp_features = temp_features[feature_cols]

    # Predict stress level
    predicted_stress = rf_model.predict(temp_features)[0]
    sensitivity_predictions.append({'daily_work_hours': hours, 'predicted_stress_level': predicted_stress})

# Convert results to a DataFrame for easy plotting
sensitivity_df = pd.DataFrame(sensitivity_predictions)

print("Sensitivity Analysis Results for Daily Work Hours:")
display(sensitivity_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.lineplot(x='daily_work_hours', y='predicted_stress_level', data=sensitivity_df, marker='o')
plt.title('Predicted Stress Level vs. Daily Work Hours (Sensitivity Analysis)')
plt.xlabel('Daily Work Hours')
plt.ylabel('Predicted Stress Level')
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(daily_work_hours_range)
plt.show()
```
