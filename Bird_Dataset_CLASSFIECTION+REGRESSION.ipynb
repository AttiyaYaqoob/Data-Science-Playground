````python
from sklearn.ensemble import RandomForestRegressor

# --- Re-establish regression data specifically for this model to ensure consistency ---
# Reload original dataset to ensure a clean state, then apply common preprocessing up to regression split
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

df_reg = pd.read_csv("attachments-1775099729337-471585475 (1).csv")

# Impute 'Interrupted_Reason'
df_reg['Interrupted_Reason'] = df_reg['Interrupted_Reason'].fillna(df_reg['Interrupted_Reason'].mode()[0])

# Drop initial set of columns
columns_to_drop_initial_reg = [
   'Tag_Battery_Level_%', 'Signal_Strength_dB', 'Tag_Type', 'Tag_Weight_g',
   'Tagged_By', 'Tracking_Quality', 'Migration_Interrupted', 'Interrupted_Reason',
   'Recovery_Location_Known', 'Recovery_Time_days', 'Observation_Counts',
   'Observation_Quality', 'Origin'
]
df_reg = df_reg.drop(columns=columns_to_drop_initial_reg)

# Create 'Calculated_Speed_kmph'
df_reg['Calculated_Speed_kmph'] = df_reg['Flight_Distance_km'] / df_reg['Flight_Duration_hours']

# Label Encode all 'object' columns
le_reg = LabelEncoder()
for col in df_reg.columns:
    if df_reg[col].dtype == 'object':
        df_reg[col] = le_reg.fit_transform(df_reg[col])

# Drop rows where 'Weather_Condition' was 'Stormy' (encoded as 4)
df_reg.drop(df_reg[df_reg['Weather_Condition'] == 4].index, inplace=True)

# Drop rows with negative values in specified columns
non_negative_cols_reg = [
    'Flight_Distance_km', 'Flight_Duration_hours', 'Average_Speed_kmph',
    'Max_Altitude_m', 'Min_Altitude_m', 'Wind_Speed_kmph', 'Humidity_%',
    'Pressure_hPa', 'Visibility_km', 'Rest_Stops', 'Predator_Sightings',
    'Flock_Size', 'Calculated_Speed_kmph'
]
for col in non_negative_cols_reg:
    if col in df_reg.columns:
        df_reg.drop(df_reg[df_reg[col] < 0].index, inplace=True)

# Drop rows where 'Flock_Size' == 499
df_reg.drop(df_reg[df_reg['Flock_Size'] == 499].index, inplace=True)

# Drop 'Bird_ID' (as it was dropped for regression in previous attempts)
df_reg = df_reg.drop('Bird_ID', axis=1)

# Define X and y for regression from this df_reg
X_reg = df_reg.drop(columns=["Flight_Duration_hours", "Average_Speed_kmph"], axis=1)
y_reg = df_reg["Flight_Duration_hours"]

# Split and scale the regression data
X_train, X_test, y_train, y_test = train_test_split(X_reg, y_reg, test_size=0.2, random_state=1000)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

X_train = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
X_test = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)
# --- End re-establishment of regression data ---

regressor = RandomForestRegressor(n_estimators=5, random_state=25)
# fit the regressor with X and Y data
regressor.fit(X_train, y_train)
# RandomForestRegressor(n_estimators=5, random_state=25)
````

### Feature Engineering: Creating New Features from Existing Ones

Feature engineering involves transforming raw data into features that better represent the underlying problem to predictive models. Since we observed weak individual correlations, combining features might create variables with stronger predictive power for `Migration_Success`.

#### Proposed New Features:

1.  **`Altitude_Range_m`**: The difference between `Max_Altitude_m` and `Min_Altitude_m`. This could indicate the variability or range of altitudes a bird experienced during migration, which might affect success.
2.  **`Migration_Efficiency_km_per_stop`**: `Flight_Distance_km` divided by `Rest_Stops`. This feature could represent how efficiently a bird migrates in terms of distance covered between rest stops. A higher value might indicate a more direct or less interrupted migration.
3.  **`Environmental_Severity_Index`**: A composite score combining `Temperature_C`, `Wind_Speed_kmph`, and `Humidity_%`. This is a more experimental feature, aiming to capture overall environmental difficulty. For simplicity, we can start with a basic combination like `Wind_Speed_kmph * Humidity_% / (Temperature_C + 1)` (adding 1 to temperature to avoid division by zero and handle negative temperatures gracefully, assuming higher values indicate harsher conditions).
4.  **`Latitude_Change`**: The absolute difference between `Start_Latitude` and `End_Latitude`. This can give a simpler measure of the north-south migration extent without longitude complexity.

````python
# 1. Altitude_Range_m
df['Altitude_Range_m'] = df['Max_Altitude_m'] - df['Min_Altitude_m']

# 2. Migration_Efficiency_km_per_stop
# Avoid division by zero: if Rest_Stops is 0, assign a very large number or handle as appropriate (e.g., impute with mean/median or set to Flight_Distance_km)
# For now, let's replace 0 with 1 to avoid Inf and interpret it as 'no rest stops, so max efficiency'
df['Migration_Efficiency_km_per_stop'] = df['Flight_Distance_km'] / df['Rest_Stops'].replace(0, 1)

# 3. Environmental_Severity_Index
# Adding a small constant to Temperature_C to prevent division by zero or negative issues
df['Environmental_Severity_Index'] = (df['Wind_Speed_kmph'] * df['Humidity_%']) / (df['Temperature_C'] + 20) # Added 20 for safety margin around 0 C

# 4. Latitude_Change
df['Latitude_Change'] = abs(df['Start_Latitude'] - df['End_Latitude'])

print("New features created successfully.")
````

We have now created the following new features:

*   `Altitude_Range_m`: Represents the vertical range covered during migration.
*   `Migration_Efficiency_km_per_stop`: Measures how many kilometers are flown per rest stop.
*   `Environmental_Severity_Index`: An attempt to capture the combined impact of wind, humidity, and temperature.
*   `Latitude_Change`: Represents the absolute change in latitude during migration.
